```python
#1
import random

a = random.randint(1, 100)
b = random.randint(1, 100)
c = random.randint(1, 100)

def Print_values(a, b, c):
    if a > b:
        if b > c:
            print(a, b, c)
        elif b <= c:
            if a > c:
                print(a, c, b)
            elif a <= c:
                print(c, a, b)
    else:
        if b > c:
            print("?")
        else:
            print(c, b, a)

print("输入的值:")
print("a =", a)
print("b =", b)
print("c =", c)

print("输出结果:")
Print_values(a, b, c)
```

```python
#2
import numpy as np

M1=np.random.randint(0,51,size=(5,10))
M2=np.random.randint(0,51,size=(10,5))

print("M1=",M1)
print("M2=",M2)

def Matrix_multip(M1, M2):
    result = []
    for i in range(5):
        row = []
        for j in range(5):
            total = 0
            for k in range(10):
                total = total + M1[i][k] * M2[k][j]
            row.append(total)
        result.append(row)
    return result

M3 = Matrix_multip(M1, M2)

print("M1 * M2 =")
print(np.array(M3))
```

```python
#3
def Pascal_triangle(k):
 
    row = [1]
    
    for i in range(k - 1):

        new_row = []
        new_row.append(1)

        for j in range(len(row) - 1):
            number = row[j] + row[j + 1]
            new_row.append(number)

        new_row.append(1)
        row = new_row
        
    print(row)

print("Pascal_triangle(100):")
Pascal_triangle(100)

print("Pascal_triangle(200):")
Pascal_triangle(200)
```

```python
#4
import random

x = random.randint(1, 100)
print("Random x =", x)

def Least_moves(x):

    moves = 0
    while x > 1:
        if x % 2 == 0:
            x = x // 2
        else:
            x = x - 1
        moves = moves + 1
    print(moves)

print("The least number of moves is:")
Least_moves(x)
```

```python
#5
import random
import matplotlib.pyplot as plt

def Find_expression(target):

    digits = "123456789"
    solutions = []

    def search(index, expression, value):

        if index == len(digits):

            if value == target:
                solutions.append(expression)

            return

        
        num = 0

        for i in range(index, len(digits)):

            num = num * 10 + int(digits[i])

            if index == 0:

                search(
                    i + 1,
                    str(num),
                    num
                )

            else:

                search(
                    i + 1,
                    expression + " + " + str(num),
                    value + num
                )

                search(
                    i + 1,
                    expression + " - " + str(num),
                    value - num
                )

    search(0, "", 0)


    for s in solutions:
        print(s, "=", target)

    print("Total solutions:", len(solutions))


number = random.randint(1,100)
print("Target number:", number)
Find_expression(number)


Total_solutions = []


for target in range(1,101):

    count = 0


    def count_search(index, value):

        global count

        digits="123456789"


        if index == len(digits):

            if value == target:
                count += 1

            return


        num = 0

        for i in range(index,len(digits)):

            num = num*10+int(digits[i])


            if index==0:

                count_search(
                    i+1,
                    num
                )

            else:

                count_search(
                    i+1,
                    value+num
                )

                count_search(
                    i+1,
                    value-num
                )


    count_search(0,0)


    Total_solutions.append(count)


print(Total_solutions)



maximum = max(Total_solutions)
minimum = min(Total_solutions)


print(
    "Maximum solutions:",
    maximum,
    "number:",
    Total_solutions.index(maximum)+1
)


print(
    "Minimum solutions:",
    minimum,
    "number:",
    Total_solutions.index(minimum)+1
)



plt.plot(
    range(1,101),
    Total_solutions
)

plt.xlabel("Number")
plt.ylabel("Number of solutions")
plt.title("Solutions from 1 to 100")
plt.show()
```
